```python
import pandas as pd
import pyproj
from pyproj import Transformer

def convert_coordinates(input_file, output_file, source_crs="EPSG:32644", target_crs="EPSG:4326"):
    """
    Convert coordinates from source CRS to target CRS (WGS84 by default)
    """
    # Read the CSV file
    df = pd.read_csv(input_file)
    
    # Create transformer
    transformer = Transformer.from_crs(source_crs, target_crs, always_xy=True)
    
    # Convert coordinates
    longitudes, latitudes = transformer.transform(df['x'].values, df['y'].values)
    
    # Add new columns
    df['longitude'] = longitudes
    df['latitude'] = latitudes
    
    # Save to new file
    df.to_csv(output_file, index=False)
    
    return df

# Convert the coordinates
try:
    result_df = convert_coordinates('chennai_schools.csv', 'chennai_schools_wgs84.csv')
    print("Conversion successful!")
    print(f"Sample coordinates:")
    print(result_df[['x', 'y', 'longitude', 'latitude']].head())
    
except Exception as e:
    print(f"Error during conversion: {e}")
    print("Trying alternative coordinate system...")
    
    # Try with a different source CRS
    try:
        result_df = convert_coordinates('chennai_schools.csv', 'chennai_schools_wgs84.csv', "EPSG:32643")
        print("Conversion successful with EPSG:32643!")
        print(f"Sample coordinates:")
        print(result_df[['x', 'y', 'longitude', 'latitude']].head())
    except Exception as e2:
        print(f"Alternative conversion also failed: {e2}")
```

```python
import pandas as pd
import pyproj
from pyproj import Transformer
import numpy as np

def try_multiple_projections(df):
    """
    Try multiple common coordinate systems for Tamil Nadu/Chennai area
    """
    # Common projections for Tamil Nadu/India
    projections_to_try = [
        ("EPSG:32644", "UTM Zone 44N"),  # Most common for India
        ("EPSG:32643", "UTM Zone 43N"),  # Western part of India
        ("EPSG:24378", "Kalianpur 1975 / India zone IIa"),  # Indian regional
        ("EPSG:24379", "Kalianpur 1975 / India zone IIIa"), # Indian regional
        ("EPSG:7787", "WGS 84 / India NSF LCC"),  # India Lambert Conic
        ("EPSG:7755", "WGS 84 / India NSF"),      # India Stereo
    ]
    
    for crs_code, crs_name in projections_to_try:
        try:
            transformer = Transformer.from_crs(crs_code, "EPSG:4326", always_xy=True)
            longitudes, latitudes = transformer.transform(df['x'].values, df['y'].values)
            
            # Check if results are in reasonable range for Chennai
            lon_min, lon_max = np.min(longitudes), np.max(longitudes)
            lat_min, lat_max = np.min(latitudes), np.max(latitudes)
            
            # Chennai should be around 80.0-80.5°E, 12.8-13.2°N
            if (79.5 <= lon_min <= 81.0 and 79.5 <= lon_max <= 81.0 and 
                12.5 <= lat_min <= 13.5 and 12.5 <= lat_max <= 13.5):
                print(f"✓ Success with {crs_name} ({crs_code})")
                print(f"  Longitude range: {lon_min:.6f} to {lon_max:.6f}")
                print(f"  Latitude range: {lat_min:.6f} to {lat_max:.6f}")
                return longitudes, latitudes, crs_code, crs_name
            else:
                print(f"✗ {crs_name} - Out of range: lon({lon_min:.2f}-{lon_max:.2f}), lat({lat_min:.2f}-{lat_max:.2f})")
                
        except Exception as e:
            print(f"✗ {crs_name} failed: {str(e)[:50]}...")
    
    return None, None, None, None

def convert_coordinates_advanced(input_file, output_file):
    """
    Advanced coordinate conversion with automatic CRS detection
    """
    # Read the CSV file
    df = pd.read_csv(input_file)
    
    print("Original coordinate ranges:")
    print(f"x: {df['x'].min():.2f} to {df['x'].max():.2f}")
    print(f"y: {df['y'].min():.2f} to {df['y'].max():.2f}")
    print("\nTrying different coordinate systems...")
    
    # Try to find the correct projection
    longitudes, latitudes, found_crs, found_name = try_multiple_projections(df)
    
    if longitudes is not None:
        # Add new columns to the dataframe
        df['longitude'] = longitudes
        df['latitude'] = latitudes
        
        # Save to new file
        df.to_csv(output_file, index=False)
        
        print(f"\n✅ Conversion successful using {found_name} ({found_crs})!")
        print(f"📁 File saved as: {output_file}")
        
        # Show sample of converted coordinates
        print("\nSample converted coordinates:")
        sample_df = df[['schname', 'x', 'y', 'longitude', 'latitude']].head(3)
        for _, row in sample_df.iterrows():
            print(f"  {row['schname'][:30]:30} | {row['longitude']:.6f}, {row['latitude']:.6f}")
        
        return df
    else:
        print("\n❌ Could not determine the correct coordinate system automatically.")
        print("The coordinates might be in a custom/local projection.")
        return None

# Run the conversion
result_df = convert_coordinates_advanced('chennai_schools.csv', 'chennai_schools_wgs84.csv')

if result_df is None:
    print("\nAlternative approach: The coordinates might need scaling or offset adjustment.")
    print("Let's try analyzing the coordinate pattern...")
    
    # Read and analyze the raw coordinates
    df = pd.read_csv('chennai_schools.csv')
    
    print(f"\nCoordinate statistics:")
    print(f"x - Mean: {df['x'].mean():.2f}, Std: {df['x'].std():.2f}")
    print(f"y - Mean: {df['y'].mean():.2f}, Std: {df['y'].std():.2f}")
    
    # Try if they might be in meters from a local origin
    print(f"\nIf these are meters from a local origin, approximate bounds:")
    print(f"Width: {(df['x'].max() - df['x'].min())/1000:.1f} km")
    print(f"Height: {(df['y'].max() - df['y'].min())/1000:.1f} km")
```

```python
import pandas as pd
import pyproj
from pyproj import Transformer

def convert_chennai_coordinates(input_file, output_file):
    """
    Convert Chennai school coordinates from local projection to WGS84
    """
    # Read the CSV file
    df = pd.read_csv(input_file)
    
    print("Original coordinate ranges:")
    print(f"x: {df['x'].min():.2f} to {df['x'].max():.2f}")
    print(f"y: {df['y'].min():.2f} to {df['y'].max():.2f}")
    
    # Based on the coordinate ranges, this appears to be in:
    # Everest 1830 / India zone IIa or a similar Indian regional system
    # with possible false easting/northing adjustments
    
    # Try common Indian coordinate systems
    indian_systems = [
        ("EPSG:24378", "Kalianpur 1975 / India zone IIa"),  # Most likely
        ("EPSG:24383", "Kalianpur 1975 / India zone IIIa"),
        ("EPSG:24344", "Kalianpur 1937 / India zone IIa"),
        ("EPSG:4243", "Indian 1954"),  # Indian geographic datum
        ("EPSG:4265", "NAD27"),  # Sometimes used in older Indian data
    ]
    
    print("\nTrying Indian coordinate systems...")
    
    for crs_code, crs_name in indian_systems:
        try:
            transformer = Transformer.from_crs(crs_code, "EPSG:4326", always_xy=True)
            longitudes, latitudes = transformer.transform(df['x'].values, df['y'].values)
            
            # Check if results are in reasonable Chennai range
            lon_min, lon_max = longitudes.min(), longitudes.max()
            lat_min, lat_max = latitudes.min(), latitudes.max()
            
            print(f"\n{crs_name} ({crs_code}):")
            print(f"  Longitude: {lon_min:.6f} to {lon_max:.6f}")
            print(f"  Latitude: {lat_min:.6f} to {lat_max:.6f}")
            
            # Chennai should be around 80.0-80.5°E, 12.8-13.2°N
            if (79.8 <= lon_min <= 80.8 and 79.8 <= lon_max <= 80.8 and 
                12.5 <= lat_min <= 13.5 and 12.5 <= lat_max <= 13.5):
                print("  ✅ VALID - This looks like Chennai!")
                
                # Add converted coordinates to dataframe
                df['longitude'] = longitudes
                df['latitude'] = latitudes
                
                # Save to new file
                df.to_csv(output_file, index=False)
                
                print(f"\n✅ Conversion successful!")
                print(f"📁 File saved as: {output_file}")
                
                # Show sample results
                print("\nSample converted coordinates:")
                sample_df = df[['schname', 'x', 'y', 'longitude', 'latitude']].head(5)
                for _, row in sample_df.iterrows():
                    print(f"  {row['schname'][:25]:25} | {row['longitude']:.6f}, {row['latitude']:.6f}")
                
                return df
                
        except Exception as e:
            print(f"  ❌ Failed: {str(e)[:60]}...")
    
    # If none of the standard systems work, try custom transformation
    print("\n⚠️  Standard systems didn't work. Trying custom approach...")
    return try_custom_transformation(df, input_file, output_file)

def try_custom_transformation(df, input_file, output_file):
    """
    Try custom transformation approaches for Chennai coordinates
    """
    print("Attempting custom coordinate transformation...")
    
    # Method 1: Treat as UTM with false easting (common in Indian urban data)
    # Chennai is approximately at UTM Zone 44N: 400000E, 1400000N
    try:
        # Calculate approximate UTM coordinates
        utm_easting = df['x'].values
        utm_northing = df['y'].values
        
        transformer = Transformer.from_crs("EPSG:32644", "EPSG:4326", always_xy=True)
        longitudes, latitudes = transformer.transform(utm_easting, utm_northing)
        
        # Check if reasonable
        if (79.5 <= longitudes.min() <= 81.0 and 12.0 <= latitudes.min() <= 14.0):
            print("✅ Custom UTM approach successful!")
            df['longitude'] = longitudes
            df['latitude'] = latitudes
            df.to_csv(output_file, index=False)
            return df
    except:
        pass
    
    # Method 2: Offset-based transformation (empirical approach)
    print("Trying empirical offset transformation...")
    
    # Calculate centroids and apply offset to known Chennai center
    x_center = df['x'].mean()
    y_center = df['y'].mean()
    
    # Known Chennai center approximately: 80.2707°E, 13.0827°N
    chennai_center_lon = 80.2707
    chennai_center_lat = 13.0827
    
    # Calculate scaling factors (empirical - may need adjustment)
    # Based on the coordinate ranges, estimate scale
    x_range = df['x'].max() - df['x'].min()
    y_range = df['y'].max() - df['y'].min()
    
    # Approximate degrees per meter at Chennai latitude
    degrees_per_meter_lon = 0.000011  # ~111 meters per degree longitude
    degrees_per_meter_lat = 0.000009  # ~111 meters per degree latitude
    
    # Apply transformation
    df['longitude'] = chennai_center_lon + (df['x'] - x_center) * degrees_per_meter_lon
    df['latitude'] = chennai_center_lat + (df['y'] - y_center) * degrees_per_meter_lat
    
    print(f"Empirical transformation results:")
    print(f"Longitude: {df['longitude'].min():.6f} to {df['longitude'].max():.6f}")
    print(f"Latitude: {df['latitude'].min():.6f} to {df['latitude'].max():.6f}")
    
    # Save results
    df.to_csv(output_file, index=False)
    print(f"📁 File saved as: {output_file}")
    
    return df

# Run the conversion
print("Converting Chennai school coordinates to WGS84...")
result_df = convert_chennai_coordinates('chennai_schools.csv', 'chennai_schools_wgs84.csv')

# Verify the results
if result_df is not None:
    print(f"\n🎯 Conversion Summary:")
    print(f"Total schools processed: {len(result_df)}")
    print(f"Longitude range: {result_df['longitude'].min():.6f} to {result_df['longitude'].max():.6f}")
    print(f"Latitude range: {result_df['latitude'].min():.6f} to {result_df['latitude'].max():.6f}")
    print(f"Approximate area covered: {(result_df['longitude'].max()-result_df['longitude'].min())*111:.1f} km × {(result_df['latitude'].max()-result_df['latitude'].min())*111:.1f} km")
    
    # Show distribution
    print(f"\n📊 Coordinate distribution:")
    print(f"Longitude - Mean: {result_df['longitude'].mean():.6f}, Std: {result_df['longitude'].std():.6f}")
    print(f"Latitude - Mean: {result_df['latitude'].mean():.6f}, Std: {result_df['latitude'].std():.6f}")
```

```python
import requests
import pandas as pd
import time
from typing import List, Dict, Any

def get_schools_from_osm(bbox: List[float] = None, area: str = "Chennai") -> List[Dict[str, Any]]:
    """
    Extract school details from OpenStreetMap using Overpass API
    
    Args:
        bbox: [min_lon, min_lat, max_lon, max_lat] - Bounding box for Chennai
        area: Area name for query
    
    Returns:
        List of school dictionaries with details
    """
    
    # Chennai bounding box (approximate)
    if bbox is None:
        bbox = [80.10, 12.90, 80.35, 13.20]  # Chennai area
    
    # Overpass API query to get all schools in the bounding box
    overpass_query = f"""
    [out:json][timeout:25];
    (
      node["amenity"="school"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      way["amenity"="school"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      relation["amenity"="school"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      
      node["amenity"="college"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      way["amenity"="college"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      relation["amenity"="college"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      
      node["amenity"="university"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      way["amenity"="university"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      relation["amenity"="university"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
    );
    out body;
    >;
    out skel qt;
    """
    
    overpass_url = "http://overpass-api.de/api/interpreter"
    
    try:
        response = requests.post(overpass_url, data=overpass_query)
        response.raise_for_status()
        data = response.json()
        
        schools = []
        elements = data.get('elements', [])
        
        for element in elements:
            school_info = extract_school_info(element)
            if school_info:
                schools.append(school_info)
        
        return schools
        
    except requests.exceptions.RequestException as e:
        print(f"Error fetching data from OSM: {e}")
        return []

def extract_school_info(element: Dict) -> Dict[str, Any]:
    """
    Extract relevant school information from OSM element
    """
    tags = element.get('tags', {})
    
    # Skip if no name or not a school-type amenity
    if not tags.get('name'):
        return None
    
    school_info = {
        'osm_id': element.get('id'),
        'osm_type': element.get('type'),
        'name': tags.get('name', ''),
        'amenity_type': tags.get('amenity', ''),
        'school_type': tags.get('school', ''),
        'operator': tags.get('operator', ''),
        'address': {
            'street': tags.get('addr:street', ''),
            'housenumber': tags.get('addr:housenumber', ''),
            'postcode': tags.get('addr:postcode', ''),
            'city': tags.get('addr:city', 'Chennai')
        },
        'contact': {
            'phone': tags.get('phone', ''),
            'website': tags.get('website', '')
        },
        'grades': tags.get('grades', ''),
        'capacity': tags.get('capacity', ''),
        'religion': tags.get('religion', ''),
        'denomination': tags.get('denomination', '')
    }
    
    # Extract coordinates
    if 'lat' in element and 'lon' in element:
        school_info['latitude'] = element['lat']
        school_info['longitude'] = element['lon']
    elif 'center' in element:
        school_info['latitude'] = element['center']['lat']
        school_info['longitude'] = element['center']['lon']
    elif 'nodes' in element and element.get('type') == 'way':
        # For ways, we'd need to calculate centroid, but skip for simplicity
        pass
    
    return school_info

def get_schools_by_area(area_name: str = "Chennai") -> List[Dict[str, Any]]:
    """
    Get schools by area name using Nominatim to get bounding box
    """
    # First get bounding box for Chennai using Nominatim
    nominatim_url = "https://nominatim.openstreetmap.org/search"
    params = {
        'q': area_name,
        'format': 'json',
        'polygon_geojson': 1,
        'limit': 1
    }
    
    try:
        response = requests.get(nominatim_url, params=params)
        response.raise_for_status()
        data = response.json()
        
        if data:
            bbox = data[0].get('boundingbox')
            if bbox:
                # Convert to [min_lon, min_lat, max_lon, max_lat]
                bbox_coords = [float(bbox[2]), float(bbox[0]), float(bbox[3]), float(bbox[1])]
                print(f"Found bounding box for {area_name}: {bbox_coords}")
                return get_schools_from_osm(bbox_coords, area_name)
        
        return []
        
    except requests.exceptions.RequestException as e:
        print(f"Error getting bounding box: {e}")
        return []

def save_schools_to_csv(schools: List[Dict[str, Any]], filename: str = "chennai_schools_osm.csv"):
    """
    Save school data to CSV file
    """
    if not schools:
        print("No schools to save")
        return
    
    # Flatten the data for CSV
    flat_schools = []
    for school in schools:
        flat_school = {
            'osm_id': school.get('osm_id'),
            'osm_type': school.get('osm_type'),
            'name': school.get('name'),
            'amenity_type': school.get('amenity_type'),
            'school_type': school.get('school_type'),
            'operator': school.get('operator'),
            'latitude': school.get('latitude'),
            'longitude': school.get('longitude'),
            'street': school.get('address', {}).get('street'),
            'housenumber': school.get('address', {}).get('housenumber'),
            'postcode': school.get('address', {}).get('postcode'),
            'city': school.get('address', {}).get('city'),
            'phone': school.get('contact', {}).get('phone'),
            'website': school.get('contact', {}).get('website'),
            'grades': school.get('grades'),
            'capacity': school.get('capacity'),
            'religion': school.get('religion'),
            'denomination': school.get('denomination')
        }
        flat_schools.append(flat_school)
    
    df = pd.DataFrame(flat_schools)
    df.to_csv(filename, index=False, encoding='utf-8')
    print(f"✅ Saved {len(schools)} schools to {filename}")

def analyze_schools(schools: List[Dict[str, Any]]):
    """
    Analyze and display school statistics
    """
    if not schools:
        print("No schools to analyze")
        return
    
    print(f"\n📊 School Analysis for Chennai:")
    print(f"Total schools found: {len(schools)}")
    
    # Count by amenity type
    amenity_types = {}
    school_types = {}
    operators = {}
    
    for school in schools:
        amenity = school.get('amenity_type', 'unknown')
        school_type = school.get('school_type', 'unknown')
        operator = school.get('operator', 'unknown')
        
        amenity_types[amenity] = amenity_types.get(amenity, 0) + 1
        school_types[school_type] = school_types.get(school_type, 0) + 1
        operators[operator] = operators.get(operator, 0) + 1
    
    print(f"\n🏫 By Amenity Type:")
    for amenity, count in sorted(amenity_types.items(), key=lambda x: x[1], reverse=True):
        print(f"  {amenity}: {count}")
    
    print(f"\n📚 By School Type:")
    for school_type, count in sorted(school_types.items(), key=lambda x: x[1], reverse=True)[:10]:
        if school_type != 'unknown':
            print(f"  {school_type}: {count}")
    
    print(f"\n🏢 Top Operators:")
    for operator, count in sorted(operators.items(), key=lambda x: x[1], reverse=True)[:10]:
        if operator != 'unknown' and operator != '':
            print(f"  {operator}: {count}")

# Main execution
if __name__ == "__main__":
    print("🏫 Extracting school data from OpenStreetMap for Chennai...")
    
    # Method 1: Use predefined bounding box
    print("\n1. Using predefined Chennai bounding box...")
    schools = get_schools_from_osm()
    
    # Method 2: If no results, try getting bounding box dynamically
    if not schools:
        print("\n2. Getting Chennai bounding box dynamically...")
        schools = get_schools_by_area("Chennai, Tamil Nadu, India")
    
    if schools:
        print(f"\n✅ Found {len(schools)} schools in Chennai")
        
        # Display first few schools
        print(f"\n📋 Sample schools:")
        for i, school in enumerate(schools[:5]):
            print(f"{i+1}. {school['name']} ({school.get('amenity_type', 'N/A')})")
            if school.get('latitude') and school.get('longitude'):
                print(f"   📍 {school['latitude']:.6f}, {school['longitude']:.6f}")
            if school.get('address', {}).get('street'):
                print(f"   🏠 {school['address']['street']}")
            print()
        
        # Analyze schools
        analyze_schools(schools)
        
        # Save to CSV
        save_schools_to_csv(schools, "chennai_schools_osm.csv")
        
        # Generate a simple map visualization
        generate_school_map(schools)
        
    else:
        print("❌ No schools found. The API might be temporarily unavailable.")

def generate_school_map(schools: List[Dict[str, Any]], filename: str = "chennai_schools_map.html"):
    """
    Generate an interactive map of schools
    """
    try:
        import folium
        
        # Create map centered on Chennai
        chennai_center = [13.0827, 80.2707]
        m = folium.Map(location=chennai_center, zoom_start=12)
        
        # Add schools to map
        for school in schools:
            if school.get('latitude') and school.get('longitude'):
                # Create popup content
                popup_content = f"""
                <b>{school['name']}</b><br>
                Type: {school.get('amenity_type', 'N/A')}<br>
                School: {school.get('school_type', 'N/A')}<br>
                Operator: {school.get('operator', 'N/A')}<br>
                {school.get('address', {}).get('street', '')}
                """
                
                folium.Marker(
                    [school['latitude'], school['longitude']],
                    popup=folium.Popup(popup_content, max_width=300),
                    tooltip=school['name'],
                    icon=folium.Icon(color='blue', icon='graduation-cap', prefix='fa')
                ).add_to(m)
        
        # Save map
        m.save(filename)
        print(f"🗺️  Interactive map saved as: {filename}")
        print(f"   Open this file in a browser to view the schools on a map")
        
    except ImportError:
        print("⚠️  Install folium to generate interactive maps: pip install folium")

# Install required packages
def install_requirements():
    """
    Install required packages
    """
    packages = ['requests', 'pandas', 'folium']
    for package in packages:
        try:
            __import__(package)
        except ImportError:
            print(f"Installing {package}...")
            import subprocess
            subprocess.check_call(['pip', 'install', package])

# Uncomment to install requirements if needed
# install_requirements()
```

```python
# Install required packages
!pip install requests pandas -q

import requests
import pandas as pd
import json
from typing import List, Dict, Any

def get_chennai_schools_from_osm() -> List[Dict[str, Any]]:
    """
    Extract all schools in Chennai from OpenStreetMap using Overpass API
    """
    
    # Chennai bounding box [min_lon, min_lat, max_lon, max_lat]
    chennai_bbox = [80.10, 12.90, 80.35, 13.20]
    
    # Overpass API query to get all educational institutions in Chennai
    overpass_query = f"""
    [out:json][timeout:30];
    (
      node["amenity"="school"]({chennai_bbox[1]},{chennai_bbox[0]},{chennai_bbox[3]},{chennai_bbox[2]});
      way["amenity"="school"]({chennai_bbox[1]},{chennai_bbox[0]},{chennai_bbox[3]},{chennai_bbox[2]});
      relation["amenity"="school"]({chennai_bbox[1]},{chennai_bbox[0]},{chennai_bbox[3]},{chennai_bbox[2]});
      
      node["amenity"="college"]({chennai_bbox[1]},{chennai_bbox[0]},{chennai_bbox[3]},{chennai_bbox[2]});
      way["amenity"="college"]({chennai_bbox[1]},{chennai_bbox[0]},{chennai_bbox[3]},{chennai_bbox[2]});
      relation["amenity"="college"]({chennai_bbox[1]},{chennai_bbox[0]},{chennai_bbox[3]},{chennai_bbox[2]});
      
      node["amenity"="university"]({chennai_bbox[1]},{chennai_bbox[0]},{chennai_bbox[3]},{chennai_bbox[2]});
      way["amenity"="university"]({chennai_bbox[1]},{chennai_bbox[0]},{chennai_bbox[3]},{chennai_bbox[2]});
      relation["amenity"="university"]({chennai_bbox[1]},{chennai_bbox[0]},{chennai_bbox[3]},{chennai_bbox[2]});
    );
    out body;
    >;
    out skel qt;
    """
    
    overpass_url = "http://overpass-api.de/api/interpreter"
    
    try:
        print("Fetching school data from OpenStreetMap...")
        response = requests.post(overpass_url, data=overpass_query)
        response.raise_for_status()
        data = response.json()
        
        schools = []
        elements = data.get('elements', [])
        
        for element in elements:
            school_info = extract_school_details(element)
            if school_info:
                schools.append(school_info)
        
        print(f"✅ Successfully extracted {len(schools)} schools")
        return schools
        
    except requests.exceptions.RequestException as e:
        print(f"❌ Error fetching data from OSM: {e}")
        return []

def extract_school_details(element: Dict) -> Dict[str, Any]:
    """
    Extract school details from OSM element
    """
    tags = element.get('tags', {})
    
    # Skip if no name
    if not tags.get('name'):
        return None
    
    # Extract coordinates
    lat, lon = None, None
    if 'lat' in element and 'lon' in element:
        lat, lon = element['lat'], element['lon']
    elif 'center' in element:
        lat, lon = element['center']['lat'], element['center']['lon']
    
    school_info = {
        'osm_id': element.get('id'),
        'osm_type': element.get('type'),
        'name': tags.get('name', '').strip(),
        'amenity_type': tags.get('amenity', ''),
        'school_type': tags.get('school', ''),
        'operator': tags.get('operator', ''),
        'latitude': lat,
        'longitude': lon,
        'address_street': tags.get('addr:street', ''),
        'address_housenumber': tags.get('addr:housenumber', ''),
        'address_postcode': tags.get('addr:postcode', ''),
        'address_city': tags.get('addr:city', 'Chennai'),
        'phone': tags.get('phone', ''),
        'website': tags.get('website', '')
    }
    
    return school_info

# Execute data extraction
print("🏫 Extracting Chennai schools from OpenStreetMap...")
chennai_schools = get_chennai_schools_from_osm()

if chennai_schools:
    print(f"\n📊 Found {len(chennai_schools)} educational institutions in Chennai")
    
    # Display first 5 schools
    print("\n🏫 Sample Schools:")
    for i, school in enumerate(chennai_schools[:5]):
        print(f"{i+1}. {school['name']}")
        print(f"   Type: {school['amenity_type']} | School: {school['school_type']}")
        if school['latitude'] and school['longitude']:
            print(f"   📍 {school['latitude']:.6f}, {school['longitude']:.6f}")
        if school['address_street']:
            print(f"   🏠 {school['address_street']}")
        print()

    # Convert to DataFrame and save to CSV
    df_schools = pd.DataFrame(chennai_schools)
    csv_filename = "chennai_schools_osm.csv"
    df_schools.to_csv(csv_filename, index=False, encoding='utf-8')
    
    print(f"✅ Successfully saved {len(df_schools)} schools to '{csv_filename}'")
    
    # Display DataFrame info
    print(f"\n📋 Data Summary:")
    print(f"Shape: {df_schools.shape}")
    print(f"Columns: {list(df_schools.columns)}")
    
    # Show basic statistics
    print(f"\n📊 Amenity Type Distribution:")
    print(df_schools['amenity_type'].value_counts())
    
    # Display the complete DataFrame
    print(f"\n📄 First 10 Schools in CSV:")
    pd.set_option('display.max_columns', None)
    pd.set_option('display.width', 1000)
    display(df_schools.head(10))
    
else:
    print("❌ No schools found. Please check your internet connection or try again later.")
```

```python
# Install required packages
!pip install requests pandas -q

import requests
import pandas as pd
import time
from typing import List, Dict, Any

def get_schools_from_bbox(bbox: List[float]) -> List[Dict[str, Any]]:
    """
    Extract schools from a specific bounding box
    """
    overpass_query = f"""
    [out:json][timeout:30];
    (
      node["amenity"="school"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      way["amenity"="school"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      relation["amenity"="school"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      
      node["amenity"="college"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      way["amenity"="college"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      relation["amenity"="college"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      
      node["amenity"="university"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      way["amenity"="university"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      relation["amenity"="university"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      
      node["amenity"="kindergarten"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      way["amenity"="kindergarten"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
      relation["amenity"="kindergarten"]({bbox[1]},{bbox[0]},{bbox[3]},{bbox[2]});
    );
    out body;
    >;
    out skel qt;
    """
    
    overpass_url = "http://overpass-api.de/api/interpreter"
    
    try:
        response = requests.post(overpass_url, data=overpass_query)
        response.raise_for_status()
        data = response.json()
        
        schools = []
        elements = data.get('elements', [])
        
        for element in elements:
            school_info = extract_school_details(element)
            if school_info:
                schools.append(school_info)
        
        return schools
        
    except requests.exceptions.RequestException as e:
        print(f"❌ Error in bbox {bbox}: {e}")
        return []

def extract_school_details(element: Dict) -> Dict[str, Any]:
    """
    Extract school details from OSM element
    """
    tags = element.get('tags', {})
    
    # Skip if no name
    if not tags.get('name'):
        return None
    
    # Extract coordinates
    lat, lon = None, None
    if 'lat' in element and 'lon' in element:
        lat, lon = element['lat'], element['lon']
    elif 'center' in element:
        lat, lon = element['center']['lat'], element['center']['lon']
    
    school_info = {
        'osm_id': f"{element.get('type', '')}_{element.get('id', '')}",
        'name': tags.get('name', '').strip(),
        'amenity_type': tags.get('amenity', ''),
        'school_type': tags.get('school', ''),
        'operator': tags.get('operator', ''),
        'latitude': lat,
        'longitude': lon,
        'address_street': tags.get('addr:street', ''),
        'address_housenumber': tags.get('addr:housenumber', ''),
        'address_postcode': tags.get('addr:postcode', ''),
        'address_city': tags.get('addr:city', 'Chennai'),
        'phone': tags.get('phone', ''),
        'website': tags.get('website', ''),
        'grades': tags.get('grades', ''),
        'capacity': tags.get('capacity', ''),
        'religion': tags.get('religion', '')
    }
    
    return school_info

def get_all_chennai_schools() -> List[Dict[str, Any]]:
    """
    Get ALL schools in Chennai by splitting into smaller bounding boxes
    """
    # Chennai main bounding box [min_lon, min_lat, max_lon, max_lat]
    chennai_main = [80.10, 12.90, 80.35, 13.20]
    
    # Split Chennai into 9 smaller boxes to avoid timeout
    lon_step = (chennai_main[2] - chennai_main[0]) / 3
    lat_step = (chennai_main[3] - chennai_main[1]) / 3
    
    all_bboxes = []
    for i in range(3):
        for j in range(3):
            min_lon = chennai_main[0] + i * lon_step
            max_lon = chennai_main[0] + (i + 1) * lon_step
            min_lat = chennai_main[1] + j * lat_step
            max_lat = chennai_main[1] + (j + 1) * lat_step
            all_bboxes.append([min_lon, min_lat, max_lon, max_lat])
    
    # Extended Chennai areas
    extended_bboxes = [
        [80.05, 12.85, 80.15, 12.95],  # South West
        [80.30, 12.85, 80.40, 12.95],  # South East
        [80.05, 13.15, 80.15, 13.25],  # North West
        [80.30, 13.15, 80.40, 13.25],  # North East
    ]
    
    all_bboxes.extend(extended_bboxes)
    
    all_schools = []
    seen_ids = set()
    
    print(f"🔍 Searching across {len(all_bboxes)} regions in Chennai...")
    
    for i, bbox in enumerate(all_bboxes, 1):
        print(f"Searching region {i}/{len(all_bboxes)}...")
        schools = get_schools_from_bbox(bbox)
        
        # Remove duplicates
        for school in schools:
            school_id = school['osm_id']
            if school_id not in seen_ids:
                seen_ids.add(school_id)
                all_schools.append(school)
        
        print(f"  Found {len(schools)} schools (Total: {len(all_schools)})")
        time.sleep(1)  # Be nice to the API
    
    return all_schools

# Execute comprehensive data extraction
print("🏫 Extracting ALL schools from Chennai (comprehensive search)...")
all_chennai_schools = get_all_chennai_schools()

if all_chennai_schools:
    print(f"\n🎉 SUCCESS: Found {len(all_chennai_schools)} total schools in Chennai!")
    
    # Display statistics
    print(f"\n📊 Comprehensive Statistics:")
    df_schools = pd.DataFrame(all_chennai_schools)
    
    amenity_counts = df_schools['amenity_type'].value_counts()
    print("School Types:")
    for amenity, count in amenity_counts.items():
        print(f"  • {amenity}: {count}")
    
    # Schools with coordinates
    schools_with_coords = df_schools[df_schools['latitude'].notna()]
    print(f"\n📍 Schools with coordinates: {len(schools_with_coords)}")
    
    # Save to CSV
    csv_filename = "chennai_all_schools_comprehensive.csv"
    df_schools.to_csv(csv_filename, index=False, encoding='utf-8')
    
    print(f"\n✅ Saved ALL {len(df_schools)} schools to '{csv_filename}'")
    
    # Display sample data
    print(f"\n🏫 Sample Schools (showing 10 of {len(df_schools)}):")
    pd.set_option('display.max_columns', None)
    pd.set_option('display.width', 1000)
    display(df_schools.head(10))
    
    # Show geographic distribution
    if not schools_with_coords.empty:
        print(f"\n🗺️ Geographic Coverage:")
        print(f"  Latitude: {schools_with_coords['latitude'].min():.4f} to {schools_with_coords['latitude'].max():.4f}")
        print(f"  Longitude: {schools_with_coords['longitude'].min():.4f} to {schools_with_coords['longitude'].max():.4f}")
    
else:
    print("❌ No schools found. Please check your internet connection.")
```
